```python
from datetime import datetime
```

```python
class DateTimeSingleton:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.datetime = datetime.now()
        return cls._instance
```

```python
DateTimeSingleton()
```

```python
program_start_on = DateTimeSingleton()
```

```python
program_start_on._instance.datetime
```

```python
type(DateTimeSingleton())
```

```python
class DateTimeSingletonInst:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = datetime.now()
        return cls._instance
```
