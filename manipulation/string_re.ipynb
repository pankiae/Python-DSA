**import the re module**

````python
import re
````

````python
paragraph = '''
John's email is john.doe123@example.com, and his phone number is 555-123-4567.
His alternate number is (555) 123-4567 and +1-555-123-4567.
His friend Jane's contact is jane_doe456@workplace.org.
An invalid email example is invalid_email@com.

They met on 2023-10-17 and also on October 17, 2023.
Another format used was 17/10/2023.
The meeting was at 2:30 PM and ended at 14:30.

The address was: 123 Main St., Apt #4B, New York, NY 10001.
The event code is ABC-987 and REF-2023-XYZ.

The ticket price was $1,200.50 or ₹5,000.

Please reach out before 2023-10-19.
Visit https://support.company.com or http://example.org.
Repeated words example: this is very very important.

<p>This is an HTML paragraph</p>
<a href="https://example.com">Click here</a>
'''
````

````python
match = re.search(r"\d{4}-\d{2}-\d{2}", paragraph)
type(match)
````

It contains:
- Match text
- Position info
- Group info
- Pattern reference

````python
print(match.group())
````

````python
match.group(1)
````

**Index over Group we have to use the parentheses in the pattern**

````python
match = re.search(r"(\d{4})-(\d{2})-(\d{2})", paragraph)
match.groups()
````

````python
print(match.group(0))
print(match.group(1))
print(match.group(2))
print(match.group(3))
````

````python
# Return the starting index where it find
print(match.start())
````

````python
# Return the ending index where it find
print(match.end())
````

````python
# Return the start and end index in tuple
print(match.span())
````

**Why Are start() and end() Important?**

In real systems:
- Syntax highlighting
- Text editors
- Data extraction pipelines
- Tokenizers
- Compiler design

They need exact position.

````python
# Returns the compiled pattern object.
match.re
# Useful for debugging.
````

````python
# Returns the original string useful for matching
match.string
````

````python
print(match.lastgroup)
````

**Internally What Happens?**

When regex engine finds match:

It creates internal structure like:

Match:
```  pattern: (\d{4})-(\d{2})-(\d{2})
  string: full paragraph
  span: (120,130)
  groups:
      0: 2023-10-17
      1: 2023
      2: 10
      3: 17
```
This object is returned.
