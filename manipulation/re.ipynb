**import the re module**

```python
import re
```

### Main functions provided by re:

| Function        | What it Does                     |
|----------------|----------------------------------|
| `re.search()`   | Finds first match                |
| `re.match()`    | Checks match at beginning        |
| `re.findall()`  | Returns all matches              |
| `re.finditer()` | Returns match objects            |
| `re.sub()`      | Replace pattern                  |
| `re.split()`    | Split string                     |
| `re.compile()`  | Compile regex for reuse          |
| `re.fullmatch()`| Matches entire string            |

### 1️⃣ Core Syntax Basic Characters:
| Pattern |   Meaning                   |
|---------|-----------------------      |
| `a`     |matches character 'a'       |
| `.`     |any character except newline |
| `\d`    |digit (0-9)                  |
| `\D`    |non-digit                    |
| `\w`    |word character (a-zA-Z0-9_ )  |
| `\W`	  |non-word character           |
| `\s`    |whitespace                   |
| `\S`    |non-whitespace               |

2️⃣ Quantifiers (Very Important in DSA):

|Pattern    |   Meaning             |
|-----------|-----------------------|
|   `*`     |   0 or more           |
|   `+`     |	1 or more           |
|   `?`     |	0 or 1              |
|   `{n}`   |exactly n times        |
|   `{n,m}` |between n and m times  |

3️⃣ Character Sets
|   Pattern |   	Meaning         |
|-----------|-----------------------|     
| `[abc]`   |   a or b or c         |     
| `[^abc]`  |   not a, b, or c      |         
| `[a-z]`   |   lowercase letters   |             
| `[A-Z]`   |   uppercase letters   |             
| `[0-9]`   |   digits              | 

4️⃣ Anchors
|   Pattern |	Meaning         |
|-----------|-------------------|
|   `^`     |	start of string |
|   `$`     |	end of string   |
|   `\b`    |	word boundary   |

5️⃣ Groups (Very Important)

Parentheses create groups:

`re.findall(r"(\d+)-(\d+)", "123-456")`

**output:**
`[('123', '456')]`

6️⃣ OR Operator

`at|dog`

Matches either "cat" or "dog".

⚙️ Flags (Advanced But Important)

|Flag   |                       |
|-------|-----------------------|
|`re.I` |	Ignore case         |
|`re.M` |	Multiline           |
|`re.S` |	Dot matches newline |
|`re.X` |	Verbose mode        |

Example:

`re.findall(r"hello", "Hello", re.I)`
